```python
df['review_lower'] = df['text'].str.lower()
```

```python
df['review_lower']
```

```python
df['is_payment'] = df['review_lower'].str.contains('refund|refunded|charge|charged|debit|debitted|money|bank|card|compensation|scam|scammed|stole|stolen|credit')
```

```python
df['is_payment'].value_counts()
```

```python
df[df['is_payment'] == True]['rating'].mean()
```

```python
df[df['is_payment'] == False]['rating'].mean()
```

```python
df['is_order_accuracy'] = df['content_lower'].str.contains('wrong|missing|didn\'t arrive|never arrived|not delivered|swapped')
df[df['is_order_accuracy'] == True]['score'].mean()
df[df['is_order_accuracy'] == False]['score'].mean()
```

```python
df['is_delivery_time'] = df['content_lower'].str.contains('late|slow|delay|delayed|waiting|hours')
df[df['is_delivery_time'] == True]['score'].mean()
df[df['is_delivery_time'] == False]['score'].mean()
```

```python
df['is_support'] = df['content_lower'].str.contains('support|customer service|customer care|agent|chat|unhelpful')
df[df['is_support'] == True]['score'].mean()
df[df['is_support'] == False]['score'].mean()
```
